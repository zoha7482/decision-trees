## Data loading

### Subtask:
Load the dataset "Position_Salaries.csv" into a pandas DataFrame.


**Reasoning**:
Load the dataset "Position_Salaries.csv" into a pandas DataFrame and display the first few rows.



```python
import pandas as pd

try:
    df = pd.read_csv('Position_Salaries.csv')
    display(df.head())
except FileNotFoundError:
    print("Error: 'Position_Salaries.csv' not found.")
    df = None
```

## Data exploration

### Subtask:
Explore the loaded dataset to understand its characteristics and identify potential issues.


**Reasoning**:
Examine the data shape, data types, descriptive statistics, and check for missing values and duplicates. Then analyze the target variable and explore the relationship between features.



```python
# Examine Data Shape
print("Data Shape:", df.shape)

# Check Data Types
print("\nData Types:\n", df.dtypes)

# Summarize Numerical Features
print("\nDescriptive Statistics:\n", df.describe())

# Analyze Target Variable
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 6))
plt.hist(df['Salary'], bins=10, color='skyblue', edgecolor='black')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.title('Distribution of Salary')
plt.show()
print(f"Range of Salary: {df['Salary'].min()} to {df['Salary'].max()}")


# Explore Relationships
plt.figure(figsize=(8, 6))
plt.scatter(df['Level'], df['Salary'], color='coral')
plt.xlabel('Level')
plt.ylabel('Salary')
plt.title('Level vs. Salary')
plt.show()

# Identify Missing Values
print("\nMissing Values:\n", df.isnull().sum())

# Handle Duplicates (If Any)
print("\nNumber of duplicate rows:", df.duplicated().sum())
```

## Data preparation

### Subtask:
Prepare the data for model training by separating the features from the target variable and handling any necessary data transformations.


**Reasoning**:
Prepare the data for model training by separating the features (Level) from the target variable (Salary).



```python
X = df['Level'].values.reshape(-1, 1)
y = df['Salary']
```

## Model training

### Subtask:
Train a Decision Tree Regression model on the prepared data.


**Reasoning**:
Train a Decision Tree Regression model on the prepared data.



```python
from sklearn.tree import DecisionTreeRegressor

# Initialize the Decision Tree Regressor
regressor = DecisionTreeRegressor()

# Fit the regressor to the data
regressor.fit(X, y)
```

## Model evaluation

### Subtask:
Evaluate the trained Decision Tree Regression model.


**Reasoning**:
Evaluate the trained Decision Tree Regression model by predicting salaries, calculating R-squared and MSE, and visualizing the results.



```python
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score, mean_squared_error

# Predict salaries using the trained model
y_pred = regressor.predict(X)

# Evaluate the model
r2 = r2_score(y, y_pred)
mse = mean_squared_error(y, y_pred)

print(f"R-squared: {r2}")
print(f"Mean Squared Error: {mse}")

# Visualize the results
plt.figure(figsize=(8, 6))
plt.scatter(y, y_pred, color='blue', label='Actual vs. Predicted')
plt.plot([min(y), max(y)], [min(y), max(y)], color='red', linestyle='--', label='Perfect Prediction')
plt.xlabel('Actual Salary')
plt.ylabel('Predicted Salary')
plt.title('Actual vs. Predicted Salaries')
plt.legend()
plt.show()
```
